```python
import gradio as gr
import numpy as np
import pandas as pd
```

```python
def cosine_similarity_my(x,y):
  dot = np.dot(x,y)
  norm_x = np.sqrt(np.sum(x**2))
  norm_y = np.sqrt(np.sum(y**2))
  cosine_similarity = dot/(norm_x*norm_y)
  return cosine_similarity
```
