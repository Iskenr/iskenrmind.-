```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Assuming 'apple' is your DataFrame

# Reset 'Brand' with random numerical data
apple['Brand'] = pd.Series(np.random.randn(len(apple)), index=apple.index)

# ... (Rest of your code to define X and y) ...

# Drop non-numeric columns before scaling
X = apple.drop(columns=['Sale Price', 'Product Name', 'Product URL', 'Ram'])

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Scale numerical features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled, y_train)
```

```python
y_preduction = model.predict(X_test_scaled)
print(y_preduction)
```
