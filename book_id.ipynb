```python
data = data[['Course Name', 'Difficulty Level','Course URL', 'Course Description', 'Skills']]
data.head()
```

```python
data['Course Name'] = data['Course Name'].str.replace(' ',',')
data['Course Name'] = data['Course Name'].str.replace(',,',',')
```

```python
data['Course Description'] = data['Course Description'].str.replace(' ','')
data['Course Description'] = data['Course Description'].str.replace(':','')
data['Course Description'] = data['Course Description'].str.replace(',','')
data['Course Description'] = data['Course Description'].str.replace(':','')
data['Skills'] = data['Skills'].str.replace('  ','')
data['Skills'] = data['Skills'].str.replace(':', '')
data['Skills'] = data['Skills'].str.replace(',','')
```

```python
data['tags'] = data['Course Name'] + data['Difficulty Level'] + data['Course Description'] + data['Skills']
data.head()
```

```python
data['tags'].iloc[1]
```

```python
data_new = data[['Course Name', 'Course URL', 'tags']]
data_new
```

```python
data_new.rename(columns = {'Course Name':'course_name', 'Course URL':'course_url'}, inplace = True)
```

```python
data_new['tags'] = data_new['tags'].apply(lambda x:x.lower())
```

```python
data_new.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
vecr = cv.fit_transform(data_new['tags']).toarray()
vars=[0]
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
  y = []
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
data_new['tags'] = data_new['tags'].apply(stem)
#
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vecr)
#
```

```python
def recommend(course):
  course_index = data_new[data_new['course_name'] == course].index[0]
  distances = similarity[course_index]
  course_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:7]
  recommended_courses = []
  for i in course_list:
    course_name = data_new.iloc[i[0]].course_name
    course_url = data_new.iloc[i[0]].course_url
    recommended_courses.append({'name':course_name,'url':course_url})
  return recommended_courses
#
```

```python
recommend('Business,Strategy:,Business,Model,Canvas,Analysis,with,Miro')
```
