```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
data = pd.read_csv('/content/data_science.csv')
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data1 = data.groupby('work_year')['salary_in_usd'].mean().sort_values(ascending=False)
data1
```

```python
data1.index
```

```python
data['avergae_salary'] = data['salary_in_usd']  + data['salary_in_usd'] * 0.2
data
```

```python
data = {
    'work_year': data1.index,
    'avergae_salary':data.values
}
data1
```

```python
data.head()
```

```python
data1 = data['avergae_salary'] = (data['avergae_salary']/100).round(2)
data1
```

```python
ten_data1 = data1.head(5)
fig, ax = plt.subplots(figsize=(10,5))
ax.bar(ten_data1.index, ten_data1.values)
ax.bar_label(ax.containers[0])
```

```python
data3 = data.job_title.value_counts()[:5]
data3
```

```python
data4 = data[['salary_in_usd', 'company_size']]
data4
```

```python
S = data4[data4['company_size']=='S']
M = data4[data4['company_size']=='M']
L = data4[data4['company_size']=='L']
print(S)
print(M)
print(L)
```

```python
lable = ['S', 'M', 'L']
```

```python
lst_mean = [S['salary_in_usd'].mean(), M['salary_in_usd'].mean(), L['salary_in_usd'].mean()]
lst_mean
```

```python
plt.figure(figsize=(10,5))
plt.bar(lable, lst_mean, color='red')
plt.xlabel('company_size')
plt.ylabel('salary_in_usd')
plt.title('company_size vs salary_in_usd')
plt.xticks(rotation= -20)
plt.show()
```
